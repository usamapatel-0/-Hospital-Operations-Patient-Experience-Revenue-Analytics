```python
#T-test — Readmission vs Average Bill
from scipy.stats import ttest_ind

readmitted_bill = df.loc[
    df["Readmission_30_Days"] == "Yes",
    "Total_Bill_INR"
]

not_readmitted_bill = df.loc[
    df["Readmission_30_Days"] == "No",
    "Total_Bill_INR"
]

t_stat, p_value = ttest_ind(
    readmitted_bill,
    not_readmitted_bill,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

```python
##ANOVA -Kya different departments ke average bills mein actual difference hai, ya sirf random variation hai?
from scipy.stats import f_oneway

groups = [
    group["Total_Bill_INR"].values
    for _, group in df.groupby("Department_Name")
]

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

```python
##Chi-square test-Kya Admission Type (Emergency/Scheduled/Referral) aur 30-day Readmission ke beech relationship hai?
from scipy.stats import chi2_contingency

table = pd.crosstab(
    df["Admission_Type"],
    df["Readmission_30_Days"]
)

chi2, p_value, dof, expected = chi2_contingency(table)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
```

```python
##Confidence Interval-Overall average hospital bill ka reasonable range kya hai?
from scipy.stats import t

mean_bill = df["Total_Bill_INR"].mean()
std_bill = df["Total_Bill_INR"].std()
n = df["Total_Bill_INR"].count()

confidence = 0.95
alpha = 1 - confidence

margin_error = (
    t.ppf(1 - alpha/2, n - 1)
    * (std_bill / np.sqrt(n))
)

lower = mean_bill - margin_error
upper = mean_bill + margin_error

print("Mean Bill:", mean_bill)
print("95% Confidence Interval:", lower, "to", upper)
```
